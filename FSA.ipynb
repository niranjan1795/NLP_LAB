```python
class DFA:
    def __init__(self):
        self.states = {'q0', 'q1', 'q2'}
        self.start_state = 'q0'
        self.accept_states = {'q1'}
        self.current_state = self.start_state

    def transition(self, char):
        if self.current_state == 'q2':
            return 'q2'

        if self.current_state == 'q0':
            if char.islower() and char.isalpha():
                self.current_state = 'q1'
            else:
                self.current_state = 'q2'
        elif self.current_state == 'q1':
            if char.islower() and char.isalpha():
                self.current_state = 'q1'
            else:
                self.current_state = 'q2'
        return self.current_state

    def reset(self):
        self.current_state = self.start_state

    def is_accepted(self, word):
        self.reset()
        if not word:
            return False
        for char in word:
            self.transition(char)
        return self.current_state in self.accept_states

    def run_check(self, word):
        if self.is_accepted(word):
            return "Accepted"
        else:
            return "Not Accepted"

    def generate_svg(self, output_path="dfa.svg"):
        svg_content = """<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 700 250" width="100%" height="250">
  <defs>
    <marker id="arrow" viewBox="0 0 10 10" refX="6" refY="5" markerWidth="6" markerHeight="6" orient="auto-start-reverse">
      <path d="M 0 1.5 L 10 5 L 0 8.5 z" fill="#4B5563" />
    </marker>
    <filter id="shadow" x="-10%" y="-10%" width="120%" height="120%">
      <feDropShadow dx="2" dy="4" stdDeviation="3" flood-opacity="0.1" flood-color="#000"/>
    </filter>
  </defs>
  <rect width="100%" height="100%" fill="#F9FAFB" rx="8" />
  <text x="350" y="35" text-anchor="middle" font-family="Inter, sans-serif" font-size="18" font-weight="bold" fill="#1F2937">DFA: Simplified English Word Recognizer</text>
  <path d="M 50 120 L 100 120" fill="none" stroke="#4B5563" stroke-width="2" marker-end="url(#arrow)" />
  <circle cx="130" cy="120" r="30" fill="#E5E7EB" stroke="#4B5563" stroke-width="2" />
  <text x="130" y="125" text-anchor="middle" font-family="Inter, sans-serif" font-size="14" font-weight="600" fill="#1F2937">q0</text>
  <path d="M 160 120 L 310 120" fill="none" stroke="#2563EB" stroke-width="2.5" marker-end="url(#arrow)" />
  <text x="240" y="110" text-anchor="middle" font-family="Inter, sans-serif" font-size="12" font-weight="bold" fill="#2563EB">[a-z]</text>
  <circle cx="340" cy="120" r="34" fill="#DBEAFE" stroke="#2563EB" stroke-width="2.5" />
  <circle cx="340" cy="120" r="28" fill="#DBEAFE" stroke="#2563EB" stroke-width="1.5" />
  <text x="340" y="125" text-anchor="middle" font-family="Inter, sans-serif" font-size="14" font-weight="bold" fill="#1E40AF">q1</text>
  <path d="M 325 93 C 315 50, 365 50, 355 93" fill="none" stroke="#2563EB" stroke-width="2" marker-end="url(#arrow)" />
  <text x="340" y="45" text-anchor="middle" font-family="Inter, sans-serif" font-size="12" font-weight="bold" fill="#2563EB">[a-z]</text>
  <path d="M 145 146 L 210 205 Q 220 215, 235 210 L 490 210 L 525 150" fill="none" stroke="#DC2626" stroke-width="1.5" stroke-dasharray="4,4" marker-end="url(#arrow)" />
  <text x="320" y="200" text-anchor="middle" font-family="Inter, sans-serif" font-size="11" fill="DC2626">other</text>
  <path d="M 370 120 L 500 120" fill="none" stroke="#DC2626" stroke-width="2" marker-end="url(#arrow)" />
  <text x="440" y="110" text-anchor="middle" font-family="Inter, sans-serif" font-size="12" font-weight="bold" fill="#DC2626">other</text>
  <circle cx="530" cy="120" r="30" fill="#FEE2E2" stroke="#DC2626" stroke-width="2" />
  <text x="530" y="125" text-anchor="middle" font-family="Inter, sans-serif" font-size="14" font-weight="600" fill="#991B1B">q2</text>
  <path d="M 515 93 C 505 50, 555 50, 545 93" fill="none" stroke="#DC2626" stroke-width="1.5" marker-end="url(#arrow)" />
  <text x="530" y="45" text-anchor="middle" font-family="Inter, sans-serif" font-size="12" fill="#DC2626">any</text>
</svg>"""
        with open(output_path, "w", encoding="utf-8") as f:
            f.write(svg_content.strip())
        print(f"Generated DFA visualization SVG at: {output_path}")

dfa_instance = DFA()
dfa_instance.generate_svg("dfa.svg")

from google.colab import files
files.download("dfa.svg")
```
